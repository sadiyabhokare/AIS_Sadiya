```python
# Function to find total number of permutations (2-letter)
def permutations(elements):
    result = []
    
    for i in range(len(elements)):
        for j in range(len(elements)):
            if i != j:  # for repetition if i != j or i==j 
                result.append(elements[i] + elements[j])
                
    return result


items = ["a", "b", "c"]
perm = permutations(items)

print("Permutations:", perm)
print("Total permutations:", len(perm))
```
