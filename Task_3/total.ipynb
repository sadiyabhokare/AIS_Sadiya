```python
# Function to find total number of combinations(2 letters)
def combinations(elements):
    result = []
    
    for i in range(len(elements)):
        for j in range(i+1,len(elements)):  # i+1 to avoid duplicates aa,bb,cc and  i < j to avoid reverse duplicates ab and ba 
            result.append(elements[i] + elements[j])
            
    return result


items = ["a", "b", "c"]
combo = combinations(items)

print("Combinations:", combo)
print("Total combinations:", len(combo))

```
