```python
# Function to count lowercase and uppercase letters
def count_upper_lower(text):
    upper = 0
    lower = 0
    
    for ch in text:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1
            
    return upper, lower


text = "STatiStiCS"
u, l = count_upper_lower(text)

print("Upper case:", u)
print("Lower case:", l)
```
