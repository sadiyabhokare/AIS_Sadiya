Array Manipulation

```python
import numpy as np
```

```python
# Multi-Dimensional Indexing
arr = np.array([[10, 20, 30],
                [40, 50, 60]])

print(arr[1, 2])
```

```python
# Masking (Conditional Indexing)
arr = np.array([10, 25, 30, 5, 40])
print(arr[arr > 20])
```

```python
# Fancy Indexing
arr = np.array([100, 200, 300, 400])

index = [0, 2]
print(arr[index])
```
