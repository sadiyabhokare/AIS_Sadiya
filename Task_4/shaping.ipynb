Array Shapes and Reshaping

```python
import numpy as np
```

```python
# Shape and Dimensions
arr = np.array([[1, 2, 3], [4, 5, 6]])

print(arr.shape)
print(arr.ndim)
```

```python
# Reshape
arr = np.arange(11, 17)
reshaped = arr.reshape(2, 3)
print(reshaped)
```

```python
# Flatten  => to convert into 1D
flat = reshaped.flatten()
print(flat)
```

```python
# Transpose
print(reshaped.transpose())
```

```python
# Concatenation
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.concatenate((a, b)))
```

```python
# Vertical & Horizontal Stack
print(np.vstack((a, b)))
print(np.hstack((a, b)))
```

```python
# Splitting
arr = np.array([10, 20, 30, 40, 50, 60])
print(np.split(arr, 3))
```
