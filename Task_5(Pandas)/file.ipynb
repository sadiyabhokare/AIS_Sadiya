File I/O

```python
import pandas as pd
```

```python
# Create DataFrame first
df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D'],
    'Age': [20, 21, 22, 23],
    'Marks': [85, 90, 78, 92]
})

print(df)
```

```python
# Write CSV File
df.to_csv('students.csv', index=False)
print("students.csv file created")
```

```python
# Read CSV File
df_csv = pd.read_csv('students.csv')
print(df_csv)
```

```python
# Write Excel File
df.to_excel('students.xlsx', index=False)
print("students.xlsx file created")
```

```python
# Read Excel File
df_excel = pd.read_excel('students.xlsx')
print(df_excel)
```
