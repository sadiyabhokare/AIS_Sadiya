Conditional Selection (AND / OR)

```python
import pandas as pd
```

```python
# Create DataFrame first
df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D'],
    'Age': [20, 21, 22, 23],
    'Marks': [85, 90, 78, 92]
})

print(df)
```

```python
# AND Condition
result_and = df[(df['Marks'] > 85) & (df['Age'] > 21)]
print(result_and)
```

```python
# OR Condition
result_or = df[(df['Marks'] > 90) | (df['Age'] < 21)]
print(result_or)
```
