Aggregation and Grouping

```python
import pandas as pd
```

```python
# GroupBy
data = {
    'Department': ['IT', 'IT', 'HR', 'HR'],
    'Salary': [50000, 60000, 40000, 45000]
}

df_group = pd.DataFrame(data)
print(df_group)
```

```python
grouped = df_group.groupby('Department')
print(grouped.mean())
```

```python
# Aggregation
print(df_group.groupby('Department').sum())
print(df_group.groupby('Department').count())
```
