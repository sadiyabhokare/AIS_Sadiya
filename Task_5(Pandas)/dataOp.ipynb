Data Operations

```python
import pandas as pd
```

```python
# Indexing & Slicing
```

```python
df = pd.DataFrame({
    'Name': ['X', 'Y'],
    'Age': [19, 20]
})
print(df)
```

```python
# Column Access
print(df['Name'])
```

```python
# Using iloc (index-based)
print(df.iloc[0])       # first row
print(df.iloc[0:2])     # first two rows
```

```python
# Using loc (label-based)
print(df.loc[0, 'Age'])
```
