Merging and Joining

```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['A', 'B']
})

df2 = pd.DataFrame({
    'ID': [1, 2],
    'Marks': [90, 85]
})
```

```python
# Merge
merged = pd.merge(df1, df2, on='ID')
print(merged)
```

```python
# Concatenation
concat_df = pd.concat([df1, df2], axis=1)
print(concat_df)
```
