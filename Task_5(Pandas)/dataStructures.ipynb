Key Data Structures

```python
import pandas as pd
```

```python
# Series (1D labeled array)
data = [10, 20, 30, 40]
s = pd.Series(data)

print(s)
```

```python
# With custom index:
s = pd.Series([100, 200, 300], index=['a', 'b', 'c'])
print(s)
```

```python
# DataFrame (2D table)
data = {
    'Name': ['A', 'B', 'C'],
    'Age': [20, 21, 22],
    'Marks': [85, 90, 88]
}

df = pd.DataFrame(data)
print(df)
```
