```python
# control statements using while loop
# break example
i = 1
while i <= 10:
    if i == 6:
        break
    print("Break example:", i)
    i = i + 1
```

```python
# continue example
i = 0
while i < 5:
    i = i + 1
    if i == 3:
        continue
    print("Continue example:", i)
```

```python
# pass example
i = 1
while i <= 3:
    pass
    i = i + 1
print("Pass example completed.")
```
