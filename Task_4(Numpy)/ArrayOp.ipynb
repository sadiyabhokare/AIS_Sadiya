Array Operations

```python
import numpy as np
```

```python
# Arithmetic Operations
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
# Broadcasting
arr = np.array([1, 2, 3])
num = 5

print(arr + num)
```

```python
# Element-wise Functions
arr = np.array([4, 9, 16])

print(np.sqrt(arr))
print(np.exp(arr))
print(np.sin(arr))
```
