Mathematical Functions

```python
import numpy as np
```

```python
# Universal Functions
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(np.add(a, b))
print(np.subtract(a, b))
print(np.multiply(a, b))
```

```python
# Statistical Functions
data = np.array([10, 20, 30, 40, 50])

print(np.mean(data))
print(np.median(data))
print(np.std(data))
print(np.var(data))
print(np.min(data))
print(np.max(data))
```

```python
# Linear Algebra
a = np.array([[1, 2],[3, 4]])

b = np.array([[5, 6],[7, 8]])

print(np.dot(a, b))

print(np.matmul(a, b))  # Alternative to np.dot for 2-D arrays 

print(np.linalg.inv(a))

print(np.linalg.eig(a))
```
