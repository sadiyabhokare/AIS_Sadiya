Arrays in Numpy

```python
# Importing the numpy library
import numpy as np
```

```python
# using np.array()
arr = np.array([10, 20, 30, 40])
print(arr)
print(type(arr))
```

```python
# using np.zeros()
zeros_arr = np.zeros(5)
print(zeros_arr)
```

```python
# using np.ones
ones_arr = np.ones((2, 3))
print(ones_arr)
```

```python
# using np.arange()
range_arr = np.arange(1, 11, 2)  # range function in python
print(range_arr)
```

```python
# np.linspace()
line_arr = np.linspace(0, 10, 5)
print(line_arr)
```

```python
# Indexing & Slicing
arr = np.array([5, 10, 15, 20, 25])

print(arr[0])      # first element
print(arr[-1])     # last element
print(arr[1:4])    # slicing
```
