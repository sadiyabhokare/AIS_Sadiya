Titanic Data Analysis

```python
# Import Libraries
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
# Load Titanic Dataset
titanic_data = sns.load_dataset('titanic')

# Display
print(titanic_data.head())
```

```python
# Basic Data Inspection

# 1. Shape (Rows, Columns)
print("Shape of the dataset:", titanic_data.shape)
```

```python
# 2. Columns
print("Column names:", titanic_data.columns)
```

```python
# 3. Data Types
print("Data types:\n", titanic_data.dtypes)
```

```python
# 4. Info
print("Dataset Info:", titanic_data.info())
```

```python
# 5. Describe
print("Dataset Description:\n", titanic_data.describe())
```

```python
# Preprocessing

# 1. Check Missing Values
print("Missing values in each column:\n", titanic_data.isnull().sum())
```

```python
# 2. fill missing 'age' with median
titanic_data['age'].fillna(titanic_data['age'].median(), inplace=True)
```

```python
# 3. Fill Embarked with mode
titanic_data['embarked'].fillna(titanic_data['embarked'].mode()[0], inplace=True)
```

```python
# 4. Drop 'deck' column due to high missing values
titanic_data.drop(columns=['deck'], inplace=True)
```

```python
# 5. Check again
print("Missing values after preprocessing:\n", titanic_data.isnull().sum())
```

```python
# Label Encoding (Convert Text → Numbers)

# 1. Check categorical columns
print(titanic_data.select_dtypes(include='object').columns)
```

```python
# 2. Encode sex
titanic_data['sex'] = titanic_data['sex'].map({'male': 0, 'female': 1})
```

```python
# 3. Encode embarked
titanic_data['embarked'] = titanic_data['embarked'].map({'S': 0, 'C': 1, 'Q': 2})
```

```python
# 4. Encode class
titanic_data['class'] = titanic_data['class'].map({'First': 1, 'Second': 2, 'Third': 3})
```

```python
# 5.Encode who
titanic_data['who'] = titanic_data['who'].map({'man': 0, 'woman': 1, 'child': 2})
```

```python
# Final Check
print("Final dataset after encoding:\n", titanic_data.head())
```
