```python
# Creating dictionary
student = {
    "name": "Sadiya",
    "age": 21,
    "course": "AI & DS"
}

# Accessing values
print(student["name"])
print(student.get("age"))

# Keys, values, items
print(student.keys())
print(student.values())
print(student.items())

# Adding / Updating
student.update({"age": 22})
student["college"] = "Engineering College"
print(student)

# Removing items
student.pop("course")
print(student)
```
