```python
a = 10
b = 5

# Arithmetic Operators
print(a + b)
print(a - b)
print(a * b)
print(a / b)

# Comparison Operators
print(a == b)
print(a != b)
print(a > b)
print(a < b)

# Logical Operators
x = True
y = False

print(x and y)
print(x or y)
print(not x)

# Assignment Operators
c = 10
c += 5
print("After += :", c)
```
