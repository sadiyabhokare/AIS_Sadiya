```python
# Creating a list
numbers = [10, 20, 30, 40, 50]

# Accessing elements
print(numbers[0])
print(numbers[1:4])

# Modifying elements
numbers[2] = 35
print(numbers)

# Adding elements
numbers.append(60)
numbers.insert(1, 15)
print(numbers)

# Removing elements
numbers.remove(35)
numbers.pop()
print(numbers)

# Other list methods
numbers.sort()
print("Sorted:", numbers)

numbers.reverse()
print("Reversed:", numbers)

extra = [70, 80]
numbers.extend(extra)
print("Extended:", numbers)

print("Count of 20:", numbers.count(20))
print("Index of 40:", numbers.index(40))

copy_list = numbers.copy()
print("Copied list:", copy_list)

numbers.clear()
print("Cleared list:", numbers)
```
